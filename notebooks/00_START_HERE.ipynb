# SemEval 2027 · Task 3 · Subtask 2
 ## Project guide
 **Question:** Does the marked occurrence of a word express the supplied sense definition?

 This folder contains the work completed so far by our two-student NLP project. Start here, then open the notebook for the step you need. Organization snapshot: **2 November 2026**.

 **Current scope:** 1,983 labeled Swedish development usages, 10 words, 705 positive labels. These are development experiments—not six-language test or leaderboard scores.

## How to work safely and reproducibly
 1. Use the Colab account that owns this Drive project (the account with Colab Pro).
 2. Read `docs/PROJECT_STATUS.md` for the current step and remaining work.
 3. Steps 3 and 4 default to **saved-score replay**. No model downloads or GPU inference occur in that mode.
 4. To intentionally run a new model experiment, change the explicit `RUN_NEW_INFERENCE` switch and use a GPU. The notebook records the resolved model revision and installed versions.
 5. Read the printed run directory. Results, chosen thresholds, predictions and provenance are saved together there.
 6. Keep candidate and threshold decisions inside the inner folds. Treat error-driven changes as exploratory until independently validated.
